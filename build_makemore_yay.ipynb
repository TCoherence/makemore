```python
xs, ys = [], []
for w in words[:1]:
    chs = ['.'] + list(w) + ['.']
    for ch1, ch2 in zip(chs, chs[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        print(ch1, ch2)
        xs.append(ix1)
        ys.append(ix2)
        
xs = torch.tensor(xs)
ys = torch.tensor(ys)
```

```python
import torch.nn.functional as F
xenc = F.one_hot(xs, num_classes=27).float()
xenc
```

```python
W = torch.randn((27, 27))
xenc @ W
```

```python
logits = (xenc @ W) # log-counts that correspond to each possible next character, but has negative values
counts = logits.exp() # counts which are all positive which is ready for normalization
probs = counts / counts.sum(1, keepdim=True) # normalization
probs
```

```python
# randomly initialize 27 neurons' weights, each neuron receive 27 inputs
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((27, 27), generator=g)
```

```python
# forward pass
xenc = F.one_hot(xs, num_classes=27).float() # input to the network: one-hot encoding
logits = xenc @ W # log-counts that correspond to each possible next character, but has negative values
counts = logits.exp() # counts which are all positive which is ready for normalization
probs = counts / counts.sum(1, keepdim=True) # normalization
# the last 2 lines here are together called "softmax!"
```
